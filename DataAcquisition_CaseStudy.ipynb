**Data Transformation**




```python
gdp_xlsx.rename(columns={'Country':'country'},inplace=True)
litrcy_rt_df.rename(columns={'name':'country'},inplace=True)
```

```python

population_csv['population'] = pd.to_numeric(population_csv['population'], errors='coerce')
gdp_xlsx['GDP'] = pd.to_numeric(gdp_xlsx['GDP'], errors='coerce')
int_usr_df['internet_users'] = pd.to_numeric(int_usr_df['internet_users'], errors='coerce')
litrcy_rt_df['literacy_rate'] = pd.to_numeric(litrcy_rt_df['literacy_rate'], errors='coerce')
```

```python
merged_df = population_csv \
    .merge(gdp_xlsx, on='country', how='outer') \
    .merge(int_usr_df, on='country', how='outer') \
    .merge(litrcy_rt_df, on='country', how='outer')

print(merged_df)
```

```python
print(f"Merged DataFrame Shape: {merged_df.shape}")
print("\nMissing values per column:")
print(merged_df.isnull().sum())
print(merged_df.head(10))
```

```python
merged_df['Internet Penetration (%)'] = ((merged_df['internet_users'] / merged_df['population']) * 100).round(2)
merged_df['GDP Per Capita'] = (merged_df['GDP'] / merged_df['population']).round(2)
print(merged_df[['country', 'population', 'GDP', 'internet_users', 'Internet Penetration (%)', 'GDP Per Capita']].head(10))
```

**Load Data**

```python
import sqlite3
conn = sqlite3.connect('world_statistics.db')

merged_df.to_sql('country_statistics', conn, if_exists='replace', index=False)

query = "SELECT * FROM country_statistics LIMIT 5;"
verification_df = pd.read_sql_query(query, conn)

conn.close()

print("Data successfully written to SQLite database!")
print(verification_df)
```

 **Data Analysis Using Pandas**

```python
top_5_gdp = merged_df.sort_values(by='GDP', ascending=False).head(5)
print("Top 5 countries by GDP:")
print(top_5_gdp[['country', 'GDP', 'population', 'GDP Per Capita']])
```

```python

averg_litrcy = merged_df['literacy_rate'].mean()
print(f"The average literacy rate across all countries is: {averg_litrcy:.2f}%")
```

```python
litrcy_rt_90_ctrys = merged_df[merged_df['literacy_rate'] > 90]

print(f"Number of countries with a literacy rate > 90%: {len(litrcy_rt_90_ctrys)}")
print(litrcy_rt_90_ctrys[['country', 'literacy_rate']])
```

```python

top_5_intr_pene = merged_df.sort_values(by='Internet Penetration (%)', ascending=False).head(5)

print("Top 5 countries with the highest internet penetration:")
print(top_5_intr_pene[['country', 'Internet Penetration (%)', 'internet_users', 'population']])
```

```python
ttl_popln = merged_df['population'].sum()
print(f"The total population is: {ttl_popln:,}")
```

```python
highst_gdp_pc = merged_df.sort_values(by='GDP Per Capita', ascending=False).iloc[0]
print(f"The country with the highest GDP per capita is {highst_gdp_pc['country']} with a GDP per capita of ${highst_gdp_pc['GDP Per Capita']:,.2f}.")
```
